```python
# Create ModelVisualizer and add robot
visualizer = ModelVisualizer(meshcat=meshcat,
                             visualize_frames=True,
                             triad_length=0.05,
                             triad_radius=0.001)
urdf_name = "humanoid_half_body.urdf"

robot_urdf_path = f"/home/tanawatp/Documents/Thesis_simulation/Models/{urdf_name}"
robot = visualizer.parser().AddModels(robot_urdf_path)
plant = visualizer.parser().plant()
robot_body_frame = plant.GetFrameByName("torso")
init_pos = [0.0,0.0,0.0] 
init_orien = np.asarray([0, 0, 0])
X_floorRobot = RigidTransform(
    RollPitchYaw(init_orien * np.pi / 180), p=[0.0,-0.0,0.0])
plant.WeldFrames(plant.world_frame(),robot_body_frame, X_floorRobot)

visualizer.Finalize()

# Create context
context = plant.CreateDefaultContext()


ind = 150
left_q = np.array([ 0.        ,  0.38810357,  0.95138131, -1.8577034 ,  0.90632209,
        0.38810357])
right_q = np.zeros(6)
q = np.concatenate((left_q,right_q))
# q = [-0.0091952,  -0.18632383,  0.31166832,  1.03721258, -1.35678643, -0.18334453,
#   0.,          0.,          0.,          0.,          0.,          0.,        ]
# q = [ 0.,         -0.20290664,  1.38388431, -1.06341363, -0.32047068, -0.20290664,
#      0.,          0.,          0.,          0.,          0.,          0.        ]
visualizer.Run(position=q,loop_once=False)
# visualizer.Run(loop_once=False)
```
